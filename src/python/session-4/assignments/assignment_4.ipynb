```python
def multiply_table(num:int)->int:
    # num=input("Enter Number")
    for i in range(1,11):
        mul=i*num
        print (f"{num} * {i} = {mul} ")
multiply_table(int(input("Enter Number")))
```

```python
def is_prime(num: int):
    for i in range(2,num):
        if num %i==0:
            return False
    return True
def twin_prime():
    for i in range(2,1001):
        if is_prime(i)==True and is_prime(i+2)==True:
            print(f"{i} and {i+2}")

twin_prime()

```

```python
def prime_factors(num:int)->list:
    out=[]
    for i in range(2,num):
        while num %i==0:
            out.append(i)
            num =num//i
    return out
prime_factors(56)
```

```python
def decimal_to_binary(num:int):
    binary = []

    while num > 0:
        remainder = num % 2
        binary.append(remainder)
        num = num // 2
    binary.reverse()

    return "".join(map(str, binary))
decimal_to_binary(11)
```

```python
def is_perfect(num: int) :
    total = 0

    for i in range(1, num):
        if num % i == 0:
            total += i

    return total == num


def perfect_numbers(start: int, end: int):
    for num in range(start, end + 1):
        if is_perfect(num):
            print(num)


perfect_numbers(0, 100)
```
