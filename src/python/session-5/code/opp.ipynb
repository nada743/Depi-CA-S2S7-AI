```python
from abc import ABC, abstractmethod
class Animal(ABC):
    @abstractmethod
    def move(self):
        pass
class Bird (Animal):
    def move(self):
        print("move from Bird")
class Cat (Animal):
        def move(self):
            print("move from cat")

a =Bird()
a.move()
                       
```

```python
class Rectangle:
    _length = 5
    width =10

    def area(self):
        return self._length * self.width
r = Rectangle()
print(r._length)
print(r.width)
print(r.area())
```

```python
class Rectangle:
    def __init__(self,length,width):
        self.__length= length
        self.__width =width
    @property
    def length(self):
        return self.__length
    @length.setter
    def length(self,value):
        if value <= 0:
            raise ValueError("length must be positive")
        self.__length =value
        
```

```python
r= Rectangle(10,20)
r.length=20
```

```python
class Model:
   
    def fit(self, x, y):
        raise NotImplementedError
    def predict(self, x):
        raise NotImplementedError
 
class LinearRegression(Model):
    def fit(self, x, y):
        print("fitting Linear regression")
   
    def predict(self, x):
        print("predicting with linear regression")
 
 
class DicisionTree(Model):
    def fit(self, x, y):
        print("fitting Decision Tree ")
   
    def predict(self, x):
        print("predicting with Decision Tree ")
       
models = [LinearRegression(), DicisionTree()]
for model in models:
    model.fit(None,None)
    model.predict(None)
```
