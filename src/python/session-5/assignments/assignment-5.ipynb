```python
class Prime:
    def __init__(self,num):
        self.num=num
    def prime_factors(self):
        out=[]
        n=self.num
        for i in range(2,n):
            while n %i==0:
                out.append(i)
                n =n//i
        return out
p=Prime(56)
p.prime_factors()
```

```python
class Binary:
    def __init__(self,num):
        self.num=num
    def decimal_to_binary(self):
        binary = []
        num=self.num
        while num > 0:
            remainder = num % 2
            binary.append(remainder)
            num = num // 2
        binary.reverse()

        return "".join(map(str, binary))
b= Binary(11)
b.decimal_to_binary()
```

```python
class Perfect:
    def __init__(self,start,end):
        self.start=start
        self.end=end
    def is_perfect(self,num) :
        total = 0

        for i in range(1, num):
            if num % i == 0:
                total += i

        return total == num


    def perfect_numbers(self):
        start=self.start
        end=self.end
        for num in range(start, end + 1):
            if self.is_perfect(num):
                print(num)
x=Perfect(0,100)
x.perfect_numbers()
```

```python
def is_perfect(num: int) :
    total = 0

    for i in range(1, num):
        if num % i == 0:
            total += i

    return total == num


def perfect_numbers(start: int, end: int):
    for num in range(start, end + 1):
        if is_perfect(num):
            print(num)


perfect_numbers(0, 100)
```
