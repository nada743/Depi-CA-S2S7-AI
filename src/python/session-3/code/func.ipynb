```python
def train_model(model_name, * datasets,**hyperparameters):
    print(f"Training Model :{model_name}")

    print("\nDatasets")
    for dataset in datasets:
        print("-",dataset)
    print("\nHyperparameters")
    for key,value in hyperparameters.items():
        print(f"{key}:{value}")
train_model(
    "Fraud Detection",
    "train.csv",
    "validtin.csv",
    "test.csv",
    learning_rate=0.001,
    epochs=100,
    batch_size="Adam",
    early_syopping=True
)
```
