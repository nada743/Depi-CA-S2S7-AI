

> ## <mark>The result above shows many missing data </mark>



###  To fix this issue You will fill these values with median for numeric and other for categorical

```python
# trying to fill missing values in my datset using for loop 
for col in df.columns:
    
    if df[col].dtype == 'object':
        df[col] = df[col].fillna('Other')

    else:
        df[col] = df[col].fillna(df[col].median())
```

```python
#check the Nulls 
df.isnull().sum()
```


# <a id="13">  13 ) . What is the  Correlation between each variable in the dataset?</a>

```python
# Make visualization to easy get insight

numeric_df = df.select_dtypes(include=['number'])

plt.figure(figsize=(14,9), dpi=200)

sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', center=0)

plt.title('Correlation Between All Numeric Variables')

plt.show()
```



> ### <mark>Your analsys and exploration here </mark>


- There is `no Correlation` for any variable with time ( `duration sec /minutes/hours`) — trip duration is independent of age, birth year, and station ids.

- `duration_sec`, `duration_min` and `duration_hr` are perfectly correlated with each other (1.0) because they are the same quantity in different units.



# <a id="14"> 14 )  .  What is the  Correlation between `specific numeric variables` in the dataset?</a>


```python
# Make visualization to easy get insight
specific_numeric = df[['duration_sec', 'duration_min', 'duration_hr', 'Age', 'member_birth_year']]

plt.figure(figsize=(10,7), dpi=200)

sns.heatmap(specific_numeric.corr(), annot=True, fmt='.2f', cmap='coolwarm', center=0)

plt.title('Correlation Between Specific Numeric Variables')

plt.show()


```

> ### <mark>Your analsys and exploration here </mark>


- There is `no Correlation` for any variable with time ( `duration sec /minutes`)




- There is a `negative Correlation` between  member_birth_year  and age which is normal (the older the birth year, the younger the person).
